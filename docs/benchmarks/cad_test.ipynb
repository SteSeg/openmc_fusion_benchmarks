```python
from jupyter_cadquery import show, set_defaults
import cadquery as cq
```

```python


set_defaults(axes=False, grid=(False, False, False), cad_width=800, height=600)

wp = cq.importers.importStep('oktavian_a.step')

# Get the underlying Shape
compound_shape = wp.val()

# Extract all solids from the compound
solids = compound_shape.Solids()

# Build an assembly
assy = cq.Assembly()
for i, solid in enumerate(solids):
    assy.add(solid, name=f"part_{i}", color=(0.6, 0.6, 0.9))
    
show(assy)

# exporters.export(assy, 'oktavian_al.glb')
```
