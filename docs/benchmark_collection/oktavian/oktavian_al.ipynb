```python
import ipywidgets as widgets
import matplotlib.pyplot as plt
from myrepo.api import get_available_results, load_result

# Dropdown to select result
available = get_available_results('oktavian')
selector = widgets.Dropdown(options=available, description="Result:")

output = widgets.Output()

def update_plot(change):
    output.clear_output()
    with output:
        data = load_result('oktavian', selector.value)
        plt.plot(data['x'], data['y'], label=selector.value)
        plt.title(f"Benchmark result: {selector.value}")
        plt.legend()
        plt.grid()
        plt.show()

selector.observe(update_plot, names='value')

# Display
display(selector, output)

# Trigger initial plot
update_plot(None)
```
