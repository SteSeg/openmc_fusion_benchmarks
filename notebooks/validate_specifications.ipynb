# Validate Specifications Against Schema

In *openmc fusion benchmarks*, the benchmark **specifications** files define a standardized set of information intended to fully describe a benchmark model. To ensure they meet repository requirements, these specifications can be validated against the **benchmark schema**.

```python
import openmc_fusion_benchmarks as ofb

ofb.validate_benchmark('oktavian_al')
```

Benchmark *specifications* validation avainst *schema* get triggered also when a benchmark gets instantiated:

```python
oktavian_al = ofb.OpenmcBenchmark('oktavian_al')
```
