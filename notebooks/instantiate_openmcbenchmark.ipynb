# Instantiate a Benchmark OpenMC Model

To instantiate a benchmark model for [OpenMC](https://docs.openmc.org/), use the `OpenmcBenchmark` class.  
At creation, *ofb* automatically validates the benchmark *specifications* against the *schema* to ensure compliance.  
If no `mesh.h5m` file is found in the working directory, *ofb* will trigger the [cad-to-dagmc](https://github.com/fusion-energy/cad_to_dagmc) meshing tool as part of the setup.

```python
import openmc_fusion_benchmarks as ofb

oktavian_al = ofb.OpenmcBenchmark(name="oktavian_al")
```

After instantiation it is possible to inspect the benchmark model via OpenMC's python api ([openmc.Model](https://docs.openmc.org/en/stable/pythonapi/generated/openmc.model.Model.html?)):

```python
model = oktavian_al.model
print(model.materials)
```

```python
print(model.tallies)
```

**Metadata** is an intersting piece of information that can be printed out of the benchmark:

```python
print(oktavian_al.metadata)
```
